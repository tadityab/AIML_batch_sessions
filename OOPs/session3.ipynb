```python
def print_decorator(func):
    def wrap():
        print("-"*20)
        func()
        print("-"*20)
    return wrap
```

```python
@print_decorator
def hello():
    print("Hello Class....")

hello()
```

```python
class Employee:
    def __init__(self, first, last):
        self.first = first 
        self.last = last 

    @property
    def email(self):
        return "{}.{}@company.com".format(self.first,self.last)

    @property
    def fullname(self):
        return "{} {}".format(self.first, self.last)

    @fullname.setter
    def fullname(self, name):
        first, last = name.split(" ")
        self.first = first
        self.last = last
        

emp_1 = Employee("Ram", "Patil")
print(emp_1.fullname)

emp_1.fullname = "Sham Jadhav"

print(emp_1.fullname)
```

```python
class Employee:
    def __init__(self, first, last):
        self.first = first 
        self.last = last 

    @property
    def email(self):
        return "{}.{}@company.com".format(self.first,self.last)

    @property
    def fullname(self):
        return "{} {}".format(self.first, self.last)

    @fullname.setter
    def fullname(self, name):
        first, last = name.split(" ")
        self.first = first
        self.last = last

    @fullname.deleter
    def fullname(self):
        print("Full name deleted")
        self.first = None
        self.last = None
        

emp_1 = Employee("Ram", "Patil")

print(emp_1.fullname)

del emp_1.fullname

print(emp_1.first)
```
